```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
from math import pi
```

```python
simulator = AerSimulator()

shots = 100
```

```python
qc_zero = QuantumCircuit(1, 1)

qc_zero.ry(pi / 3, 0)

qc_zero.measure(0, 0)

qc_zero.draw("mpl")
```

```python
compiled_zero = transpile(qc_zero, simulator)

result_zero = simulator.run(
    compiled_zero,
    shots=100
).result()

counts_zero = result_zero.get_counts()

print("Measurement results:", counts_zero)
```

```python
plot_histogram(
    counts_zero,
    title="Biased Quantum Coin - More Zeros"
)
```

```python
qc_one = QuantumCircuit(1, 1)

qc_one.x(0)

qc_one.ry(pi / 3, 0)

qc_one.measure(0, 0)

qc_one.draw("mpl")
```

```python
compiled_one = transpile(qc_one, simulator)

result_one = simulator.run(
    compiled_one,
    shots=100
).result()

counts_one = result_one.get_counts()

print("Measurement results:", counts_one)
```

```python
plot_histogram(
    counts_one,
    title="Biased Quantum Coin - More Ones"
)
```

```python
plot_histogram(
    [counts_zero, counts_one],
    legend=["More Zeros", "More Ones"],
    title="Challenge 2 - Biased Quantum Coin Comparison"
)
```
