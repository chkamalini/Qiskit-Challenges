```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
```

```python
simulator = AerSimulator()

shots = 100
```

```python
qc_a = QuantumCircuit(1, 1)

qc_a.x(0)

qc_a.measure(0, 0)

qc_a.draw("mpl")
```

```python
compiled_a = transpile(qc_a, simulator)

result_a = simulator.run(compiled_a, shots=100).result()

counts_a = result_a.get_counts()

print("Circuit A Results:", counts_a)
```

```python
plot_histogram(counts_a, title="Circuit A: One X Gate")
```

```python
qc_b = QuantumCircuit(1, 1)

qc_b.x(0)
qc_b.x(0)

qc_b.measure(0, 0)

qc_b.draw("mpl")
```

```python
compiled_b = transpile(qc_b, simulator)

result_b = simulator.run(compiled_b, shots=100).result()

counts_b = result_b.get_counts()

print("Circuit B Results:", counts_b)
```

```python
plot_histogram(counts_b, title="Circuit B: Two X Gates")
```

```python
qc_c = QuantumCircuit(1, 1)

qc_c.h(0)

qc_c.measure(0, 0)

qc_c.draw("mpl")
```

```python
compiled_c = transpile(qc_c, simulator)

result_c = simulator.run(compiled_c, shots=100).result()

counts_c = result_c.get_counts()

print("Circuit C Results:", counts_c)
```

```python
plot_histogram(counts_c, title="Circuit C: One H Gate")
```

```python
print("Circuit A - One X Gate:", counts_a)
print("Circuit B - Two X Gates:", counts_b)
print("Circuit C - One H Gate:", counts_c)
```

```python
plot_histogram(
    [counts_a, counts_b, counts_c],
    legend=["Circuit A: One X", "Circuit B: Two X", "Circuit C: One H"],
    title="Challenge 1: Comparison of Measurement Results"
)
```
